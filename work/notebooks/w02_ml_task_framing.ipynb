## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
import pandas as pd

df = pd.read_csv("content_refresh_anonymized.csv")

# Unit of analysis: one row = one content item
filtered = df[(df["impressions_90d"] > 0) & (df["content_age_days"] >= 90)].copy()

# Sketch the target column
filtered["is_declining"] = (filtered["trend_direction"] == "down").astype(int)

# Show the unit of analysis as an actual dataframe
display_cols = ["content_id", "impressions_90d", "sessions_90d", "content_age_days",
                 "trend_direction", "is_declining"]
print(filtered[display_cols].head(10))

print(f"\nShape: {filtered.shape[0]:,} rows x {filtered.shape[1]} columns")
print(f"Target distribution:\n{filtered['is_declining'].value_counts(normalize=True)}")
```

## 5. Why ML beats a fixed rule here

Last week's numbers already showed this directly: the stale_visible_page reason code, a single fixed rule, only caught 17 rows out of 30,000 (0.1%), while declining_with_demand caught 13,152 (43.8%). Two rules built on the same data disagree wildly depending on which signals they check. A fixed rule can only look at one or two conditions at a time; a model can weigh many weak, partially-overlapping signals together (impressions, position, freshness, engagement) and find combinations no single hand-written rule would catch. The starter pipeline's own results confirm this isn't theoretical: the baseline rule scored 0.240 Precision@50, while a random forest using the same underlying signals scored 0.740, more than triple the hit rate on the same review capacity.
